```python
# notebooks/mushroom_frequent_itemset_mining.ipynb

import sys
sys.path.append('../src')

# Step 1: 导入必要的库
import pandas as pd
from src.mushroom_preprocessing import load_dataset, preprocess_data
from src.frequent_itemset_mining import mine_frequent_itemsets

# Step 2: 加载和预处理 Mushroom 数据集
df = load_dataset('data/mushroom.data')
df_transformed = preprocess_data(df)

# Step 3: 挖掘频繁项集和关联规则
min_support = 0.1
frequent_itemsets_apriori, frequent_itemsets_fpgrowth, rules_apriori, rules_fpgrowth = mine_frequent_itemsets(df_transformed, min_support)

# Step 4: 显示频繁项集
print(frequent_itemsets_apriori.head())
print(frequent_itemsets_fpgrowth.head())

# Step 5: 结果可视化
import matplotlib.pyplot as plt

# 绘制频繁项集的支持度分布
frequent_itemsets_apriori['support'].plot(kind='bar', figsize=(10, 6))
plt.title('Apriori Frequent Itemsets Support Distribution')
plt.xlabel('Itemsets')
plt.ylabel('Support')
plt.show()
```
